# Разведочный анализ данных. Исследование и визуализация данных.

## 1) Текстовое описание набора данных

В качестве набора данных будем использовать набор данных 
по диабету из библиотеки scikit-learn - https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_diabetes.html

Этот датасет представляет собой медицинские данные о пациентах с диабетом. 
Задача заключается в предсказании количественной меры прогрессирования диабета 
через год на основе различных физиологических и биохимических показателей.

Датасет содержит следующие признаки:
- age - возраст пациента
- sex - пол пациента
- bmi - индекс массы тела
- bp - среднее артериальное давление
- s1 - общий холестерин
- s2 - липопротеины низкой плотности
- s3 - липопротеины высокой плотности
- s4 - общий холестерин / липопротеины высокой плотности 
- s5 - логарифм уровня триглицеридов 
- s6 - уровень глюкозы в крови 
- target - количественная мера прогрессирования диабета через год (целевая переменная)

Датасет содержит 442 наблюдения.

# Импорт библиотек
Импортируем библиотеки с помощью команды import. Как правило, все команды import размещают в первой ячейке ноутбука, но мы в этом примере будем подключать все библиотеки последовательно, по мере их использования.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
%matplotlib inline 
sns.set(style="ticks")
```

# Загрузка данных

Загрузим датасет с помощью функции load_diabetes из библиотеки scikit-learn. 
Эта функция возвращает объект Bunch, который содержит данные (data), целевые значения (target) 
и названия признаков (feature_names).

```python
# Загружаем датасет diabetes
diabetes_data = load_diabetes()

# Создаем DataFrame из данных
data = pd.DataFrame(diabetes_data.data, columns=diabetes_data.feature_names)

# Добавляем целевую переменную
data['target'] = diabetes_data.target
```

# 3) Визуальное исследование датасета
Для визуального исследования могут быть использованы различные виды диаграмм.

### Диаграмма рассеяния

Позволяет построить распределение двух колонок данных и визуально обнаружить наличие зависимости. Не предполагается, что значения упорядочены (например, по времени).

```python
fig, ax = plt.subplots(figsize=(6,6)) 
sns.scatterplot(ax=ax, x='s5', y='target', data=data)
plt.xlabel('Логарифм уровня триглицеридов')
plt.ylabel('Прогрессирование диабета')
plt.title('Зависимость прогрессирования диабета от уровня триглицеридов')
plt.show()
```

# 4) Информация о корреляции признаков
Проверка корреляции признаков позволяет решить две задачи:
1. Понять какие признаки (колонки датасета) наиболее сильно коррелируют с целевым признаком (в нашем примере это колонка "target"). Именно эти признаки будут наиболее информативными для моделей машинного обучения. Признаки, которые слабо коррелируют с целевым признаком, можно попробовать исключить из построения модели, иногда это повышает качество модели. Нужно отметить, что некоторые алгоритмы машинного обучения автоматически определяют ценность того или иного признака для построения модели.
1. Понять какие нецелевые признаки линейно зависимы между собой. Линейно зависимые признаки, как правило, очень плохо влияют на качество моделей. Поэтому если несколько признаков линейно зависимы, то для построения модели из них выбирают какой-то один признак.

```python
data.corr()
```

Корреляционная матрица содержит коэффициенты корреляции между всеми парами признаков.

Корреляционная матрица симметрична относительно главной диагонали. На главной диагонали расположены единицы (корреляция признака самого с собой).

На основе корреляционной матрицы можно сделать следующие выводы:
- Целевой признак наиболее сильно коррелирует с признаками bmi и s5.
- Наблюдается сильная положительная корреляция между признаками s1 и s2 (примерно 0.9).
- Признаки s3 и s4 имеют сильную отрицательную корреляцию (примерно -0.74).
- Признаки s4 и s5 демонстрируют умеренную положительную корреляцию (примерно 0.62).

По умолчанию при построении матрицы используется коэффициент корреляции Пирсона. Возможно также построить корреляционную матрицу на основе коэффициентов корреляции Кендалла и Спирмена. На практике три метода редко дают значимые различия.

```python
data.corr(method='pearson')
```

```python
data.corr(method='kendall')
```

```python
data.corr(method='spearman')
```

### Визуализация корреляционной матрицы

Тепловая карта (heatmap) позволяет визуально представить корреляционную матрицу.

```python
sns.heatmap(data.corr())
plt.title('Корреляционная матрица признаков')
plt.tight_layout()
plt.show()
```

```python
sns.heatmap(data.corr(), annot=True, fmt='.2f')
plt.title('Корреляционная матрица признаков')
plt.tight_layout()
plt.show()
```

```python
sns.heatmap(data.corr(), cmap='YlGnBu', annot=True, fmt='.2f')
plt.title('Корреляционная матрица признаков')
plt.tight_layout()
plt.show()
```

```python
# Создаем маску для верхнего треугольника
mask = np.triu(np.ones_like(data.corr(), dtype=bool))
sns.heatmap(data.corr(), mask=mask, annot=True, fmt='.2f')
plt.title('Корреляционная матрица признаков в треугольной форме')
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 3, sharex='col', sharey='row', figsize=(20,5))
sns.heatmap(data.corr(method='pearson'), ax=ax[0], annot=True, fmt='.2f')
ax[0].set_title('Pearson')
sns.heatmap(data.corr(method='kendall'), ax=ax[1], annot=True, fmt='.2f')
ax[1].set_title('Kendall')
sns.heatmap(data.corr(method='spearman'), ax=ax[2], annot=True, fmt='.2f')
ax[2].set_title('Spearman')
plt.suptitle('Сравнение методов корреляции')
plt.tight_layout()
plt.show()
```

### Анализ корреляции признаков с целевой переменной

Выведем корреляции всех признаков с целевой переменной в порядке убывания.

```python
# Вычисляем корреляции с целевой переменной
correlations = data.corr()['target'].drop('target').sort_values(ascending=False)

print('Корреляции признаков с целевой переменной (по убыванию):')
for feature, corr_value in correlations.items():
    print('{}: {:.4f}'.format(feature, corr_value))
```

```python
# Визуализация корреляций с целевой переменной
plt.figure(figsize=(6,4))
correlations.plot(kind='barh')
plt.xlabel('Корреляция с прогрессированием диабета')
plt.title('Корреляция признаков с целевой переменной')
plt.tight_layout()
plt.show()
```
